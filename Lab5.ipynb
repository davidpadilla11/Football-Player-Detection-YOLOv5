```python
import torch
import cv2
```

```python
model = torch.hub.load('ultralytics/yolov5', 'custom', path='yolov5/runs/train/exp4/weights/best.pt')
```

```python
video_path = 'TOP 10 GOALS EURO 2024.mp4'
output_path = 'video_salidaEuro.mp4'
```

```python
cap = cv2.VideoCapture(video_path)
```

```python
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = int(cap.get(cv2.CAP_PROP_FPS))
```

```python
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_path, fourcc, fps, (width, height))
```

```python
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    results = model(frame)

    for det in results.xyxy[0].numpy():
        x1, y1, x2, y2, conf, cls = map(int, det)
        label = f'{model.names[cls]} {conf:.2f}'
        cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2)
        cv2.putText(frame, label, (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

    out.write(frame)
```

```python
cap.release()
out.release()
print("El video se ha procesado y guardado correctamente.")
```
